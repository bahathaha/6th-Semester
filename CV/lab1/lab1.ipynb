```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
```

```python
path = "IRIS.csv"
df = pd.read_csv(path)
```

```python
df.head()
```

```python
fig, axes = plt.subplots(2, 2, figsize=(10,10))
df.boxplot(by='species', column=['sepal_length'], grid=False, ax=axes[0,0])
df.boxplot(by='species', column=['sepal_width'], grid=False, ax=axes[0,1])
df.boxplot(by='species', column=['petal_length'], grid=False, ax=axes[1,0])
df.boxplot(by='species', column=['petal_width'], grid=False, ax=axes[1,1])
plt.show()
```

```python
fig, axes = plt.subplots(2, 3, figsize=(15, 10))
species = df["species"].unique()

# Plot 1
for sp in species:
    subset = df[df["species"] == sp]
    axes[0, 0].scatter(subset["sepal_length"], subset["sepal_width"], label=sp)
axes[0, 0].set_xlabel("sepal length")
axes[0, 0].set_ylabel("sepal width")
axes[0, 0].set_title("Sepal Length vs Sepal Width")
axes[0, 0].legend()

# Plot 2
for sp in species:
    subset = df[df["species"] == sp]
    axes[0, 1].scatter(subset["petal_length"], subset["petal_width"], label=sp)
axes[0, 1].set_xlabel("petal length")
axes[0, 1].set_ylabel("petal width")
axes[0, 1].set_title("Petal Length vs Petal Width")
axes[0, 1].legend()

# Plot 3
for sp in species:
    subset = df[df["species"] == sp]
    axes[0, 2].scatter(subset["sepal_length"], subset["petal_length"], label=sp)
axes[0, 2].set_xlabel("sepal length")
axes[0, 2].set_ylabel("petal length")
axes[0, 2].set_title("Sepal Length vs Petal Length")
axes[0, 2].legend()

# Plot 4
for sp in species:
    subset = df[df["species"] == sp]
    axes[1, 0].scatter(subset["sepal_length"], subset["petal_width"], label=sp)
axes[1, 0].set_xlabel("sepal length")
axes[1, 0].set_ylabel("petal width")
axes[1, 0].set_title("Sepal Length vs Petal Width")
axes[1, 0].legend()

# Plot 5
for sp in species:
    subset = df[df["species"] == sp]
    axes[1, 1].scatter(subset["petal_length"], subset["sepal_width"], label=sp)
axes[1, 1].set_xlabel("petal length")
axes[1, 1].set_ylabel("sepal width")
axes[1, 1].set_title("Petal Length vs Sepal Width")
axes[1, 1].legend()

# Plot 6
for sp in species:
    subset = df[df["species"] == sp]
    axes[1, 2].scatter(subset["sepal_width"], subset["petal_width"], label=sp)
axes[1, 2].set_xlabel("sepal width")
axes[1, 2].set_ylabel("petal width")
axes[1, 2].set_title("Sepal Width vs Petal Width")
axes[1, 2].legend()

plt.show()

```

```python
fig, axes = plt.subplots(2, 2, figsize=(12, 10))

def qq_plot(data, ax, title):
    data = np.sort(data)
    n = len(data)

    # Generate normal data with same mean & std
    normal_data = np.random.normal(
        loc=np.mean(data),
        scale=np.std(data),
        size=n
    )
    normal_data = np.sort(normal_data)

    ax.scatter(normal_data, data)
    ax.plot(normal_data, normal_data)
    ax.set_title(title)
    ax.set_xlabel("Theoretical Quantiles")
    ax.set_ylabel("Sample Quantiles")

qq_plot(df["sepal_length"], axes[0, 0], "Q-Q Plot: Sepal Length")
qq_plot(df["sepal_width"], axes[0, 1], "Q-Q Plot: Sepal Width")
qq_plot(df["petal_length"], axes[1, 0], "Q-Q Plot: Petal Length")
qq_plot(df["petal_width"], axes[1, 1], "Q-Q Plot: Petal Width")

plt.tight_layout()
plt.show()
```
